# Yeast Protein Localization

Decision Tree classification with 5-fold stratified cross-validation.

```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.model_selection import StratifiedKFold, cross_val_predict
from sklearn.metrics import (
    f1_score, recall_score, matthews_corrcoef,
    roc_auc_score, accuracy_score, classification_report
)
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv("yeast.csv")

X = data.drop("name", axis=1)
y = data["name"]
```

```python
model = DecisionTreeClassifier(
    criterion="gini",
    max_depth=5,
    min_samples_leaf=5,
    random_state=42
)
```

```python
cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)
```

```python
pred = cross_val_predict(model, X, y, cv=cv)

prob = cross_val_predict(
    model, X, y, cv=cv, method="predict_proba"
)
```

```python
model.fit(X, y)

print("Accuracy:", accuracy_score(y, pred))
print("Macro F1 Score:", f1_score(y, pred, average="macro"))
print("Macro Recall:", recall_score(y, pred, average="macro"))
print("MCC:", matthews_corrcoef(y, pred))

print("Macro AUC:", roc_auc_score(
    y, prob,
    labels=model.classes_,
    multi_class="ovr",
    average="macro"
))

print("\nClassification Report:")
print(classification_report(y, pred, zero_division=0))
```

```python
plt.figure(figsize=(22, 10))
plot_tree(
    model,
    feature_names=X.columns,
    class_names=model.classes_,
    filled=True,
    rounded=True,
    fontsize=8
)
plt.show()
```
